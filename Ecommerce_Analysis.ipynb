```python

from pyspark.sql import SparkSession

# Create Spark Session
spark = SparkSession.builder \
    .appName("Ecommerce Product Analysis") \
    .getOrCreate()

# Load CSV File
df = spark.read.csv(
    r"C:\Users\User\Desktop\ecommerce_bigdata_dataset.csv",
    header=True,
    inferSchema=True
)

# Show Data
df.show(30)
```

```python
df.printSchema()
```

```python
df.count()
```

```python
df.orderBy(df.Price.desc()).show(20)
```

```python
df.groupBy("Category").count().show()
```

```python
from pyspark.sql.functions import avg

df.groupBy("Category").agg(avg("Price")).show()
```

```python
df.orderBy(df.Rating.desc()).show(25)
```

```python
df.filter(df.Stock < 5000).show()
```
